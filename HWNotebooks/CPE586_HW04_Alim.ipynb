# Q1 - Locally-weighted Regression

### Given

The weighted least squares problem is

$$
\hat{\mathbf{w}}=
\arg\min_{\mathbf{w}}
\left[
\frac{1}{2}\sum_{i=1}^{N}a^{(i)}
\left(y^{(i)}-\mathbf{w}^T\mathbf{x}^{(i)}\right)^2
+\frac{\lambda}{2}\|\mathbf{w}\|^2
\right]
$$

We need to show that

$$
{
\hat{\mathbf{w}}=(X^TAX+\lambda I)^{-1}X^TA\mathbf{y}
}
$$

where $X$ is the $N\times D$ design matrix, $A$ is a diagonal matrix with $A_{ii}=a^{(i)}$, and $I$ is the identity matrix.


**Step 1: Express the loss function in matrix form.**

The given loss function is

$$
J(\mathbf{w})=
\frac{1}{2}\sum_{i=1}^{N}a^{(i)}
\left(y^{(i)}-\mathbf{w}^T\mathbf{x}^{(i)}\right)^2
+\frac{\lambda}{2}\|\mathbf{w}\|^2
$$

Since $A$ is a diagonal matrix containing the weights, the loss function can be written as

$$
J(\mathbf{w})=
\frac{1}{2}(\mathbf{y}-X\mathbf{w})^T
A(\mathbf{y}-X\mathbf{w})
+\frac{\lambda}{2}\mathbf{w}^T\mathbf{w}
$$

**Step 2: Expand the loss function.**

Expanding the quadratic term,

$$
J(\mathbf{w})=
\frac{1}{2}\left[
\mathbf{y}^TA\mathbf{y}
-\mathbf{y}^TAX\mathbf{w}
-\mathbf{w}^TX^TA\mathbf{y}
+\mathbf{w}^TX^TAX\mathbf{w}
\right]
+\frac{\lambda}{2}\mathbf{w}^T\mathbf{w}
$$

Since $A$ is symmetric and the cross terms are equal,

$$
J(\mathbf{w})=
\frac{1}{2}\mathbf{y}^TA\mathbf{y}
-\mathbf{w}^TX^TA\mathbf{y}
+\frac{1}{2}\mathbf{w}^TX^TAX\mathbf{w}
+\frac{\lambda}{2}\mathbf{w}^T\mathbf{w}
$$

**Step 3: Differentiate with respect to $\mathbf{w}$.**

Taking the gradient,

$$
\frac{\partial J}{\partial\mathbf{w}}
=
-X^TA\mathbf{y}
+X^TAX\mathbf{w}
+\lambda\mathbf{w}
$$

Therefore,

$$
{
\nabla_{\mathbf{w}}J
=
X^TAX\mathbf{w}-X^TA\mathbf{y}
+\lambda\mathbf{w}
}
$$

**Step 4: Set the gradient equal to zero.**

For minimization,

$$
\nabla_{\mathbf{w}}J=0
$$

Thus,

$$
X^TAX\mathbf{w}-X^TA\mathbf{y}
+\lambda\mathbf{w}=0
$$

Rearranging,

$$
X^TAX\mathbf{w}+\lambda\mathbf{w}
=X^TA\mathbf{y}
$$

Factoring out $\mathbf{w}$,

$$
(X^TAX+\lambda I)\mathbf{w}=X^TA\mathbf{y}
$$

**Step 5: Solve for $\mathbf{w}$.**

Assuming $X^TAX+\lambda I$ is invertible, multiplying both sides by its inverse gives

$$
{
\hat{\mathbf{w}}=
(X^TAX+\lambda I)^{-1}X^TA\mathbf{y}
}
$$

$$

\left[

# Q3 - How is Logistics Regression different from Polynomial Regression?

Polynomial regression is used to predict continuous numerical values by modeling a nonlinear relationship between the input features and the output. Logistic regression is primarily used for classification, particularly binary classification, by predicting the probability of belonging to a class. Polynomial regression typically minimizes squared error, while logistic regression minimizes cross-entropy and produces probabilities between 0 and 1.

### Mathematical Models

**Polynomial Regression:**

$$
\hat{y}=w_0+w_1x+w_2x^2+\cdots+w_kx^k
$$

**Logistic Regression:**

$$
h_\theta(x)=\frac{1}{1+e^{-\theta^Tx}}
$$


# Q4 - Assumptions and Diagnostics in Linear Regression

### Part 1: How will you detect the violation of the homoscedasticity assumption?

Homoscedasticity means that the variance of residuals remains constant across different values of the predictors.

To detect its violation:

1. Fit the linear regression model and calculate the residuals.
2. Create a residual plot against the predictors or fitted values.
3. Examine the spread of the residuals:
   - **Homoscedasticity:** Residuals have approximately constant spread throughout the plot.
   - **Heteroscedasticity:** Residuals show increasing or decreasing spread, often forming a funnel or fan-shaped pattern.
4. A scale-location plot can also help identify changes in residual spread.


### Part 2: How will you detect the violation of the normality of errors assumption?

The normality assumption states that regression errors follow a normal distribution.

To detect its violation:

1. **Residual histogram:** Plot the distribution of residuals. An approximately symmetric, bell-shaped distribution supports the normality assumption.
2. **Q-Q plot:** Compare the residual quantiles with theoretical normal distribution quantiles.
   - Points approximately following the reference straight line indicate normality.
   - Significant deviations from the line suggest non-normality.
3. The Shapiro-Wilk test can also be used to check normality.

**Conclusion:** A non-bell-shaped residual distribution or substantial deviations in the Q-Q plot indicate possible violation of normality.

### Part 3: In multiple regression, how will you detect multicollinearity?

Multicollinearity occurs when two or more predictor variables are highly correlated.

To detect multicollinearity:

1. **Scatterplot matrix:** Plot the predictors against one another to identify strong linear relationships.
2. **Correlation matrix:** Examine correlations between predictors. Values close to +1 or -1 indicate strong pairwise correlation.
3. **Variance Inflation Factor (VIF):** Calculate the VIF for each predictor.
   - VIF below 5: Low multicollinearity.
   - VIF between 5 and 10: Moderate multicollinearity.
   - VIF above 10: Severe multicollinearity.



### Part 4 (Bonus): What statistical test will you use to determine whether errors are dependent?

The **Durbin-Watson test** is used to detect first-order autocorrelation between consecutive regression errors.

**Mathematical framework:**

The test statistic is

$$
DW=\frac{\sum_{i=2}^{N}(e_i-e_{i-1})^2}
{\sum_{i=1}^{N}e_i^2}
$$

where $e_i$ is the residual for the $i$-th observation and $N$ is the total number of observations.

**Hypotheses:**

- $H_0$: No first-order autocorrelation exists.
- $H_1$: First-order autocorrelation exists.

**Procedure and interpretation:**

1. Fit the regression model and calculate the residuals in their original order.
2. Calculate the Durbin-Watson statistic.
3. Interpret the result:
   - $DW\approx2$: No first-order autocorrelation is suggested.
   - $DW<2$: Positive autocorrelation is suggested.
   - $DW>2$: Negative autocorrelation is suggested.
4. For a formal hypothesis test, compare the statistic with the appropriate Durbin-Watson critical values.



### Part 5 (Bonus): What statistical test will you use to determine whether errors are non-normal?

The **Shapiro-Wilk test** is used to determine whether regression residuals follow a normal distribution.

**Mathematical framework:**

The Shapiro-Wilk test statistic is

$$
W=\frac{\left(\sum_{i=1}^{N}a_i e_{(i)}\right)^2}
{\sum_{i=1}^{N}(e_i-\bar{e})^2}
$$

where:

- $e_{(i)}$ represents the ordered residuals.
- $\bar{e}$ is the mean of the residuals.
- $a_i$ represents coefficients derived from the expected order statistics and covariance of a normal sample.
- $N$ is the number of observations.

**Hypotheses:**

- $H_0$: The errors follow a normal distribution.
- $H_1$: The errors do not follow a normal distribution.

**Procedure and interpretation:**

1. Fit the regression model and calculate the residuals.
2. Apply the Shapiro-Wilk test using **stats.shapiro(residuals)**, as demonstrated in the lecture notebook.
3. The test returns a W statistic and a p-value.
4. At a significance level of 0.05:
   - If $p<0.05$, reject $H_0$, indicating evidence of non-normality.
   - If $p\geq0.05$, fail to reject $H_0$, indicating insufficient evidence of non-normality.
5. A Q-Q plot can also be used to visually assess normality.

```python

import torch
import torch.nn as nn
import numpy as np

class CauchyRegression(nn.Module):

    def __init__(self, input_dim, c=1.0):
        super().__init__()

        self.linear = nn.Linear(
            input_dim, 1, dtype=torch.float64
        )
        self.c = c
        self.loss_history = []

    def forward(self, X):
        return self.linear(X)

    def cauchy_loss(self, y_pred, y_true):
        residual = (y_true - y_pred) / self.c

        loss = (self.c**2 / 2) * torch.log1p(
            residual**2
        )
        return torch.mean(loss)

    def fit(self, X, y, learning_rate=1.0, epochs=3000):

        X = torch.tensor(X, dtype=torch.float64)
        y = torch.tensor(
            y.reshape(-1, 1), dtype=torch.float64
        )

        # Standardize input features
        self.X_mean = X.mean(dim=0)
        self.X_std = X.std(dim=0, unbiased=False)
        X_scaled = (X - self.X_mean) / self.X_std

        # Initialize parameters
        with torch.no_grad():
            self.linear.weight.zero_()
            self.linear.bias.fill_(y.mean().item())

        # Optimizer
        optimizer = torch.optim.SGD(
            self.parameters(), lr=learning_rate
        )

        self.loss_history = []

        # Training loop
        for epoch in range(epochs):

            y_pred = self.forward(X_scaled)
            loss = self.cauchy_loss(y_pred, y)

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            self.loss_history.append(loss.item())

            if (epoch + 1) % 500 == 0:
                print(
                    f"Epoch {epoch+1}/{epochs}, "
                    f"Loss: {loss.item():.6f}"
                )

        with torch.no_grad():
            self.final_loss = self.cauchy_loss(
                self.forward(X_scaled), y
            ).item()

        return self

    def predict(self, X):

        X = torch.tensor(X, dtype=torch.float64)
        X_scaled = (X - self.X_mean) / self.X_std

        with torch.no_grad():
            return self.forward(X_scaled).numpy().flatten()

    def get_coefficients(self):

        with torch.no_grad():
            w = self.linear.weight.flatten() / self.X_std
            w0 = self.linear.bias.item() - torch.dot(
                w, self.X_mean
            ).item()

        return w0, w.numpy()
```

```python

# Select features and target
X = df[['AT', 'V', 'AP', 'RH']].values
y = df['PE'].values

# Initialize the model
model = CauchyRegression(input_dim=4, c=1.0)

# Train the model
model.fit(
    X, y,
    learning_rate=1.0,
    epochs=3000
)

print("\nTraining completed.")
```

A 95% confidence interval provides a range of plausible values for the true population regression coefficient. If the model assumptions and confidence interval method are valid, 95% of intervals constructed from repeated samples would contain the true coefficient.

The confidence intervals help interpret the fitted regression model:

1. **Statistical significance:** If the 95% confidence interval excludes zero, the corresponding coefficient is statistically significant at the 5% level under a standard two-sided test.

2. **Direction of relationship:** An interval entirely above zero indicates a positive relationship, while an interval entirely below zero indicates a negative relationship, holding other predictors constant.

3. **Precision:** A narrower confidence interval indicates a more precise coefficient estimate, while a wider interval indicates greater uncertainty.

4. **Interval containing zero:** If the confidence interval includes zero, there is insufficient evidence to conclude that the corresponding coefficient differs from zero at the 5% significance level.

```python

# Q5 Part 7: Cauchy Regression using Statsmodels

import numpy as np
import pandas as pd
import statsmodels.api as sm
from statsmodels.base.model import GenericLikelihoodModel

# Select features and target
features = ['AT', 'V', 'AP', 'RH']

X_original = df[features].astype(float)
y_sm = df['PE'].astype(float)

# Standardize features for numerical stability
X_mean = X_original.mean()
X_std = X_original.std(ddof=0)

X_scaled = (X_original - X_mean) / X_std

# Add intercept
X_sm = sm.add_constant(X_scaled, has_constant='add')


# Custom Cauchy Model
class CauchyModel(GenericLikelihoodModel):

    def __init__(self, endog, exog, c=1.0, **kwargs):
        self.c = c
        super().__init__(endog, exog, **kwargs)

    def nloglikeobs(self, params):

        # Predicted values
        y_pred = np.dot(self.exog, params)

        # Residuals
        residual = self.endog - y_pred

        # Cauchy negative log-likelihood
        loss = (
            np.log(np.pi * self.c)
            + np.log1p((residual / self.c)**2)
        )

        return loss
```

```python

# Initial coefficients using least squares
start_params = np.linalg.lstsq(
    X_sm.to_numpy(),
    y_sm.to_numpy(),
    rcond=None
)[0]

# Create CauchyModel
cauchy_sm = CauchyModel(
    y_sm,
    X_sm,
    c=1.0
)

# Fit using maximum likelihood estimation
results_sm = cauchy_sm.fit(
    start_params=start_params,
    method='bfgs',
    maxiter=1000,
    disp=False
)

# Display results
print(results_sm.summary())

print("\nOptimization converged:",
      results_sm.mle_retvals['converged'])
```

```python

# Convert statsmodels coefficients to original units
params = np.asarray(results_sm.params)

sm_weights = params[1:] / X_std.to_numpy()

sm_intercept = (
    params[0]
    - np.sum(sm_weights * X_mean.to_numpy())
)

sm_coefficients = np.r_[sm_intercept, sm_weights]

# PyTorch coefficients
torch_intercept, torch_weights = model.get_coefficients()
torch_coefficients = np.r_[torch_intercept, torch_weights]

# Compare the coefficients
comparison = pd.DataFrame({
    'Parameter': ['w0', 'w1', 'w2', 'w3', 'w4'],
    'Variable': ['Intercept', 'AT', 'V', 'AP', 'RH'],
    'PyTorch': torch_coefficients,
    'Statsmodels': sm_coefficients
})

display(comparison.round(4))

# Calculate comparable mean Cauchy loss
y_pred_sm = np.asarray(X_sm @ results_sm.params)

cauchy_loss_sm = np.mean(
    0.5 * np.log1p((y_sm.to_numpy() - y_pred_sm)**2)
)

print(f"PyTorch Final Loss: {model.final_loss:.6f}")
print(f"Statsmodels Cauchy Loss: {cauchy_loss_sm:.6f}")
```


### Custom Logistic Regression Using PyTorch

A custom binary logistic regression model is implemented using PyTorch.

The sigmoid function converts the linear model output into a probability:

$$
h_\theta(x)=\frac{1}{1+e^{-(w^Tx+b)}}
$$

The model is trained by minimizing binary cross-entropy loss using stochastic gradient descent (SGD).

The input features are standardized using the training data statistics. The original training and testing datasets are kept separate.


```python

# Custom logistic regression class

class LogisticRegressionSGD(nn.Module):

    def __init__(self, input_dim, learning_rate=0.05,
                 epochs=200, batch_size=128):
        super().__init__()

        self.linear = nn.Linear(input_dim, 1)
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.batch_size = batch_size
        self.loss_history = []

    def forward(self, X):
        # Return linear scores (logits)
        return self.linear(X)

    def fit(self, X, y):

        torch.manual_seed(42)

        X = torch.as_tensor(X, dtype=torch.float32)
        y = torch.as_tensor(
            y, dtype=torch.float32
        ).reshape(-1, 1)

        dataset = TensorDataset(X, y)
        loader = DataLoader(
            dataset,
            batch_size=self.batch_size,
            shuffle=True
        )

        loss_fn = nn.BCEWithLogitsLoss()

        optimizer = torch.optim.SGD(
            self.parameters(),
            lr=self.learning_rate
        )

        self.loss_history = []

        for epoch in range(self.epochs):

            self.train()
            total_loss = 0.0

            for X_batch, y_batch in loader:

                logits = self.forward(X_batch)
                loss = loss_fn(logits, y_batch)

                optimizer.zero_grad()
                loss.backward()
                optimizer.step()

                total_loss += loss.item() * len(X_batch)

            avg_loss = total_loss / len(dataset)
            self.loss_history.append(avg_loss)

            if (epoch + 1) % 25 == 0:
                print(
                    f"Epoch {epoch+1}/{self.epochs}, "
                    f"Loss: {avg_loss:.6f}"
                )

        return self

    def predict_proba(self, X):

        self.eval()
        X = torch.as_tensor(X, dtype=torch.float32)

        with torch.no_grad():
            probabilities = torch.sigmoid(self.forward(X))

        return probabilities.numpy().flatten()

    def predict(self, X, threshold=0.5):

        probabilities = self.predict_proba(X)
        return (probabilities >= threshold).astype(int)
```

```python

# Train custom logistic regression

torch.manual_seed(42)

logistic_model = LogisticRegressionSGD(
    input_dim=4,
    learning_rate=0.05,
    epochs=200,
    batch_size=128
)

logistic_model.fit(X_train, y_train)

print("\nTraining completed.")
print(f"Final Training Loss: {logistic_model.loss_history[-1]:.6f}")
```
