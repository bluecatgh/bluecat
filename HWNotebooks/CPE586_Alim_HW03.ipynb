#Q3 - Constrained Optimization using Lagrange Multiplier

# 1. Set up the Lagrangian

Given

$$
f(x,y,z)=x^2+y^2+z^2
$$

subject to

$$
x+2y+z=6.
$$

The constraint can be written as

$$
g(x,y,z)=x+2y+z-6=0.
$$

The Lagrangian is

$$
{
L(x,y,z,\lambda)
=
x^2+y^2+z^2-\lambda(x+2y+z-6)
}
$$

# 2. Derive the System of Equations

The partial derivatives of $L$:

$$
\frac{\partial L}{\partial x}
=
2x-\lambda=0
$$

$$
\frac{\partial L}{\partial y}
=
2y-2\lambda=0
$$

$$
\frac{\partial L}{\partial z}
=
2z-\lambda=0
$$

$$
\frac{\partial L}{\partial \lambda}
=
-(x+2y+z-6)=0.
$$

Therefore, the system of equations is

$$
{
2x-\lambda=0
}
$$

$$
{
2y-2\lambda=0
}
$$

$$
{
2z-\lambda=0
}
$$

$$
{
x+2y+z=6
}
$$

# 3. Solve for the Critical Point

From

$$
2x-\lambda=0,
$$

we get

$$
x=\frac{\lambda}{2}.
$$

From

$$
2y-2\lambda=0,
$$

we get

$$
y=\lambda.
$$

From

$$
2z-\lambda=0,
$$

we get

$$
z=\frac{\lambda}{2}.
$$

Substitute these into the constraint:

$$
\frac{\lambda}{2}
+
2\lambda
+
\frac{\lambda}{2}
=
6.
$$

Therefore,

$$
3\lambda=6
$$

and

$$
\lambda=2.
$$

Hence,

$$
x=1,\qquad y=2,\qquad z=1.
$$

Therefore, the critical point is

$$
{(x,y,z)=(1,2,1)}.
$$

# 4. Classify the Extrema

$$

# Q5 - Linear Regression and Regularization

The linear regression problem is

$$
\underset{w}{\arg\min}
\left\{
\|y-Xw\|^2+\lambda\|w\|^2
\right\}.
$$

Using

$$
\|v\|^2=v^Tv,
$$

the objective function can be written as

$$
J(w)
=
(y-Xw)^T(y-Xw)
+
\lambda w^Tw.
$$

### Step 1: Expand the Objective Function

Expanding the first term,

$$
(y-Xw)^T(y-Xw)
=
y^Ty-y^TXw-w^TX^Ty+w^TX^TXw.
$$

Since

$$
y^TXw=w^TX^Ty,
$$

we obtain

$$
J(w)
=
y^Ty
-
2w^TX^Ty
+
w^TX^TXw
+
\lambda w^Tw.
$$

### Step 2: Take the Gradient with Respect to $w$

Taking the gradient,

$$
\frac{\partial J}{\partial w}
=
-2X^Ty
+
2X^TXw
+
2\lambda w.
$$

For the minimum, set the gradient equal to zero:

$$
-2X^Ty
+
2X^TXw
+
2\lambda w
=
0.
$$

Dividing by $2$,

$$
-X^Ty
+
X^TXw
+
\lambda w
=
0.
$$

Therefore,

$$
X^TXw+\lambda w=X^Ty.
$$

Since

$$
\lambda w=\lambda Iw,
$$

we can write

$$
(X^TX+\lambda I)w=X^Ty.
$$

### Step 3: Solve for $\hat{w}$

Multiplying both sides by

$$
(X^TX+\lambda I)^{-1},
$$

we obtain

$$
{
\hat{w}
=
(X^TX+\lambda I)^{-1}X^Ty
}
$$

This is the solution for linear regression with an $L_2$ penalty.

### Comparison with the Original Linear Regression Solution

For ordinary linear regression without the $L_2$ penalty,

$$
\lambda=0.
$$

Therefore,

$$
\hat{w}_{OLS}
=
(X^TX)^{-1}X^Ty.
$$

With the $L_2$ penalty,

$$
\hat{w}_{L2}
=
(X^TX+\lambda I)^{-1}X^Ty.
$$

Hence, the difference is that the regularized solution adds

$$
\lambda I
$$

to

$$
X^TX.
$$

Therefore,

$$
{
\hat{w}_{OLS}
=
(X^TX)^{-1}X^Ty
}
$$

and

$$
{
\hat{w}_{L2}
=
(X^TX+\lambda I)^{-1}X^Ty
}
$$

### Impact of Adding the $L_2$ Penalty

The penalty term is

$$
\lambda\|w\|^2.
$$

When

$$
\lambda=0,
$$

the solution becomes the ordinary linear regression solution.

As $\lambda$ increases, larger values of the weights are penalized more strongly. Therefore, the estimated weights are pushed toward smaller values.

Thus, the $L_2$ penalty helps control the magnitude of the model weights and can reduce overfitting.

In summary,

$$
{
\text{larger }\lambda
\rightarrow
\text{stronger penalty on large weights}
}
$$

and

$$
{
\lambda=0
\rightarrow
\text{ordinary least-squares regression}
}
$$

# Q6 - Assumptions in Linear Regression

For the model

$$
y_i=w_0+w_1x_i+\epsilon_i,
$$

the assumptions are:

1. **Linear relationship**

$$
E[y_i]=w_0+w_1x_i.
$$

The mean response is linear in the predictor $x$.

2. **Zero-mean error**

$$
E[\epsilon_i]=0.
$$

3. **Constant error variance (homoscedasticity)**

$$
Var(\epsilon_i)=\sigma^2
$$

for every observation.

4. **Errors are uncorrelated**

$$
Cov(\epsilon_i,\epsilon_j)=0,\qquad i\neq j.
$$

5. **The predictor values are treated as fixed/non-random** in the
simple linear regression derivation.

6. **Normal-error assumption**

For statistical inference, the errors are assumed to follow a normal
distribution,

$$
\epsilon_i\sim N(0,\sigma^2).
$$

# Q7 - Computing Derivative using PyTorch Autograd

```python
import torch

# Values
r = torch.tensor(30000.1, dtype=torch.float64, requires_grad=True)
rs = torch.tensor(0.00887, dtype=torch.float64)

# Function
tau = 2 * r * torch.sqrt(r / rs) * torch.sqrt(1 - rs / r)

# Compute derivative d(tau)/dr
tau.backward()

print("tau =", tau.item())
print("d(tau)/dr =", r.grad.item())
```

# Q8 - Linear Regression in One Variable with Scikit-Learn

A simple linear regression model is created using **steer angle** as the
independent variable and **yaw rate** as the response variable.

The linear regression model is

$$
\hat{y}=w_0+w_1x
$$

where $w_0$ is the intercept and $w_1$ is the coefficient of steer angle.

The dataset is divided into training and test sets, with $25\%$ of the
data used for testing.

```python
import numpy as np
import matplotlib.pyplot as plt
import polars as pl

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split


# Load the RAV4 dataset
url = "https://raw.githubusercontent.com/rahulbhadani/Datasets/master/RAV4/2020-04-15-05-12-00_2T3Y1RFV8KC014025_CAN_Messages_5_Hz.csv"

data = pl.read_csv(url)


# Remove rows containing missing values
data = data.drop_nulls()

data.head()


# Select the independent and response variables
# Independent variable (X): steer angle
# Response variable (y): yaw rate
y = data['yaw_rate']
X = data[['steer_angle']]


# Split the dataset into training and test sets
# 25% of the data is used for testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)


# Create and train the linear regression model
linearModel = LinearRegression()

linearModel.fit(X_train, y_train)


# Obtain the estimated model coefficients
print("w0 (Intercept):", linearModel.intercept_)
print("w1 (Coefficient):", linearModel.coef_[0])


# Compute R^2 using the test data
r_squared = linearModel.score(X_test, y_test)

print("R^2 (Coefficient of Determination):", r_squared)


# Report the fitted linear regression model and final results
print("Estimated Linear Regression Model:")
print(f"Yaw Rate = {linearModel.intercept_:.6f} + "
      f"{linearModel.coef_[0]:.6f} × Steer Angle")

print(f"\nw0 (Intercept) = {linearModel.intercept_:.6f}")
print(f"w1 (Coefficient) = {linearModel.coef_[0]:.6f}")
print(f"Test R^2 = {r_squared:.6f}")
```

### Results

$$

$$

=
-0.207197
+
0.066208(\text{Steer Angle})
}
$$

The coefficient of determination computed using the test data is

}
